```
# %%
import os, json, warnings, sys
import numpy as np, pandas as pd
import matplotlib
if os.environ.get("HEADLESS"): matplotlib.use("Agg")
import matplotlib.pyplot as plt
import seaborn as sns
warnings.filterwarnings("ignore")

try:                                  # Colab only
    get_ipython().system("pip install xgboost optuna shap statsmodels lightgbm catboost -q")
except Exception:
    pass
import optuna, shap
from xgboost import XGBClassifier
from sklearn.model_selection import (train_test_split, StratifiedKFold, RepeatedStratifiedKFold,
                                     cross_val_score, cross_val_predict)
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import MinMaxScaler, StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
from sklearn.tree import DecisionTreeClassifier
from sklearn.isotonic import IsotonicRegression
from sklearn.calibration import calibration_curve
from sklearn.metrics import (accuracy_score, precision_score, recall_score, f1_score, roc_auc_score,
                             brier_score_loss, confusion_matrix, roc_curve)
from statsmodels.stats.contingency_tables import mcnemar
from scipy.stats import spearmanr
optuna.logging.set_verbosity(optuna.logging.WARNING)

SEED, N_TRIALS, K_FOLDS, N_BOOT = 42, int(os.environ.get("N_TRIALS", 50)), 5, 1000
TARGET = "Heart_Risk"
CSV = os.environ.get("CSV", "heart_disease_risk_dataset_earlymed (1).csv")
OUT = os.environ.get("OUT", "results"); os.makedirs(OUT, exist_ok=True)

# unified figure style (one font, one size, one palette)
plt.rcParams.update({"font.family": "DejaVu Sans", "font.size": 11, "axes.titlesize": 12, "axes.labelsize": 11,
                     "legend.fontsize": 10, "figure.dpi": 120, "savefig.dpi": 300, "savefig.bbox": "tight",
                     "axes.spines.top": False, "axes.spines.right": False})
BLUE, RED, GREY = "#1f4e79", "#b03a2e", "#7f8c8d"
def save(name): plt.savefig(f"{OUT}/{name}.png"); plt.show()
R = {"seed": SEED, "n_trials": N_TRIALS, "cv_folds": K_FOLDS}

# %% [markdown]
# ## 1. Load data, duplicates BEFORE splitting, class balance

# %%
if not os.path.exists(CSV):
    try:
        from google.colab import files; files.upload()
    except Exception: pass
df = pd.read_csv(CSV)
print(df.shape); print(df.dtypes); print("Missing:", int(df.isnull().sum().sum()))
R["n_raw"] = len(df); R["class_before"] = df[TARGET].value_counts().to_dict()
R["n_duplicates"] = int(df.duplicated().sum())
df = df.drop_duplicates().dropna().reset_index(drop=True)
R["n_after_dedup"] = len(df); R["class_after"] = df[TARGET].value_counts().to_dict()
print(f"Duplicates removed: {R['n_duplicates']}  | rows kept: {len(df)}")
print("Class balance before:", R["class_before"], " after:", R["class_after"])
if not pd.api.types.is_numeric_dtype(df["Gender"]):
    df["Gender"] = df["Gender"].astype("category").cat.codes
feats = [c for c in df.columns if c != TARGET]
# same predictors with different labels? (label noise / determinism check)
g = df.groupby(feats)[TARGET].nunique()
R["unique_feature_vectors"] = int(len(g)); R["vectors_with_conflicting_labels"] = int((g > 1).sum())
print("Unique predictor vectors:", len(g), "| with conflicting labels:", int((g > 1).sum()))

# %% [markdown]
# ## 2. ONE stratified split (seed 42). Test set is touched once, at the end.

# %%
X, y = df[feats], df[TARGET].values
Xdev, Xte, ydev, yte = train_test_split(X, y, test_size=0.20, stratify=y, random_state=SEED)
R.update(n_dev=len(ydev), n_test=len(yte), dev_pos_neg=[int(ydev.sum()), int((1-ydev).sum())],
         test_pos_neg=[int(yte.sum()), int((1-yte).sum())])
print("Development:", Xdev.shape, R["dev_pos_neg"], "| Test:", Xte.shape, R["test_pos_neg"])
skf = StratifiedKFold(K_FOLDS, shuffle=True, random_state=SEED)         # SAME folds in every trial
num_cols = ["Age"]
def pipe(est):  # scaling is fitted on training folds only
    return Pipeline([("pre", ColumnTransformer([("age", MinMaxScaler(), num_cols)], remainder="passthrough")), ("clf", est)])
def xgb(params): return XGBClassifier(**params, objective="binary:logistic", eval_metric="logloss",
                                      random_state=SEED, n_jobs=-1, tree_method="hist")

# %% [markdown]
# ## 3. EDA (kept from your notebook, shortened) and target-leakage diagnostics

# %%
plt.figure(figsize=(5, 4)); sns.countplot(x=df[TARGET], color=BLUE); plt.title("Target class distribution"); save("fig_target")
plt.figure(figsize=(9, 7)); sns.heatmap(df.corr(numeric_only=True), cmap="RdYlBu", center=0); plt.title("Correlation matrix"); save("fig_corr")
auc1 = {c: roc_auc_score(ydev, Xdev[c]) for c in feats}
R["single_feature_auc_top"] = dict(sorted(auc1.items(), key=lambda kv: -abs(kv[1]-.5))[:6])
R["depth3_tree_cv_auc"] = float(cross_val_score(DecisionTreeClassifier(max_depth=3, random_state=SEED), Xdev, ydev, cv=skf, scoring="roc_auc").mean())
print("Top single-feature AUCs:", R["single_feature_auc_top"]); print("Depth-3 tree CV AUC:", R["depth3_tree_cv_auc"])

# %% [markdown]
# ## 4. Bayesian optimization (Optuna TPE) on the development set only

# %%
SPACE = {"n_estimators": (100, 500, "int"), "max_depth": (3, 10, "int"), "learning_rate": (0.01, 0.3, "float"),
         "subsample": (0.6, 1.0, "float"), "colsample_bytree": (0.6, 1.0, "float"), "gamma": (0.0, 5.0, "float"),
         "min_child_weight": (1, 10, "int")}          # your original ranges
R["search_space"] = SPACE
def objective(trial):
    p = {k: (trial.suggest_int(k, lo, hi) if t == "int" else trial.suggest_float(k, lo, hi)) for k, (lo, hi, t) in SPACE.items()}
    return cross_val_score(pipe(xgb(p)), Xdev, ydev, cv=skf, scoring="accuracy").mean()
study = optuna.create_study(direction="maximize", sampler=optuna.samplers.TPESampler(seed=SEED), pruner=optuna.pruners.NopPruner())
study.optimize(objective, n_trials=N_TRIALS)
best = study.best_params; R["best_params"] = best; R["best_cv_accuracy"] = study.best_value
R["default_xgb_cv_accuracy"] = float(cross_val_score(pipe(xgb({})), Xdev, ydev, cv=skf, scoring="accuracy").mean())
print("Best CV accuracy:", study.best_value, "| default XGBoost CV accuracy:", R["default_xgb_cv_accuracy"]); print(best)
tr = pd.DataFrame({"trial": [t.number for t in study.trials], "cv_accuracy": [t.value for t in study.trials]}); tr.to_csv(f"{OUT}/optuna_trials.csv", index=False)
plt.figure(figsize=(6, 4)); plt.scatter(tr.trial, tr.cv_accuracy, s=14, color=BLUE, label="Trial"); plt.plot(tr.trial, tr.cv_accuracy.cummax(), color=RED, label="Best so far")
plt.xlabel("Trial"); plt.ylabel(f"{K_FOLDS}-fold CV accuracy"); plt.title("Optimization history"); plt.legend(); save("fig_optuna_history")
imp_o = optuna.importance.get_param_importances(study)
plt.figure(figsize=(6, 4)); plt.barh(list(imp_o)[::-1], list(imp_o.values())[::-1], color=BLUE); plt.xlabel("Importance"); plt.title("Hyperparameter importance"); save("fig_param_importance")

# %% [markdown]
# ## 5. Threshold and calibrators from OUT-OF-FOLD development predictions (no test data)

# %%
final = pipe(xgb(best))
oof = cross_val_predict(final, Xdev, ydev, cv=skf, method="predict_proba")[:, 1]
fpr_o, tpr_o, th_o = roc_curve(ydev, oof); J = tpr_o - fpr_o; k = int(np.argmax(J)); T_STAR = float(th_o[k])
R["threshold_oof_youden"] = T_STAR; print("T* (from out-of-fold dev predictions):", T_STAR)
lg = lambda p: np.log(np.clip(p, 1e-6, 1-1e-6) / (1-np.clip(p, 1e-6, 1-1e-6))).reshape(-1, 1)
platt = LogisticRegression(C=1e6).fit(lg(oof), ydev); iso = IsotonicRegression(out_of_bounds="clip").fit(oof, ydev)
plt.figure(figsize=(6, 4)); plt.plot(th_o[1:], J[1:], color=BLUE, lw=2); plt.axvline(T_STAR, color=RED, ls="--", label=f"T* = {T_STAR:.4f}")
plt.xlabel("Probability threshold"); plt.ylabel("Youden's J"); plt.title("Threshold selection (out-of-fold, development set)"); plt.legend(); save("fig_threshold")

# %% [markdown]
# ## 6. Final model, ONE evaluation on the untouched test set

# %%
def ece_score(y, p, n_bins=10):
    idx = np.clip(np.digitize(p, np.linspace(0, 1, n_bins+1)[1:-1]), 0, n_bins-1); e = 0.0
    for m in range(n_bins):
        s = idx == m
        if s.any(): e += s.mean() * abs(y[s].mean() - p[s].mean())
    return float(e)
def metrics(y, p, thr):
    yh = (p >= thr).astype(int); tn, fp, fn, tp = confusion_matrix(y, yh).ravel()
    return dict(accuracy=accuracy_score(y, yh), precision=precision_score(y, yh), recall=recall_score(y, yh), specificity=tn/(tn+fp),
                f1=f1_score(y, yh), auc=roc_auc_score(y, p), brier=brier_score_loss(y, p), ece=ece_score(y, p), tn=int(tn), fp=int(fp), fn=int(fn), tp=int(tp))
def boot_ci(y, p, thr, n=N_BOOT):
    rng = np.random.default_rng(SEED); rows = []
    for _ in range(n):
        i = rng.integers(0, len(y), len(y))
        if len(np.unique(y[i])) == 2:
            m = metrics(y[i], p[i], thr); rows.append({k: m[k] for k in ["accuracy","precision","recall","specificity","f1","auc","brier","ece"]})
    d = pd.DataFrame(rows); return {c: (float(d[c].quantile(.025)), float(d[c].quantile(.975))) for c in d}

final.fit(Xdev, ydev); p_te = final.predict_proba(Xte)[:, 1]
m05, mT = metrics(yte, p_te, 0.5), metrics(yte, p_te, T_STAR)
print("TEST @0.50 :", {k: round(v, 5) for k, v in m05.items()}); print("TEST @T*   :", {k: round(v, 5) for k, v in mT.items()})
R["test_at_0.5"], R["test_at_Tstar"], R["ci_test_at_Tstar"] = m05, mT, boot_ci(yte, p_te, T_STAR)
print("95% CIs @T*:", R["ci_test_at_Tstar"])
a1, a2 = (p_te >= .5).astype(int) == yte, (p_te >= T_STAR).astype(int) == yte
tab = [[int((a1&a2).sum()), int((a1&~a2).sum())], [int((~a1&a2).sum()), int((~a1&~a2).sum())]]
R["mcnemar_0.5_vs_Tstar"] = {"table": tab, "p_value": float(mcnemar(tab, exact=True).pvalue)}; print("McNemar:", R["mcnemar_0.5_vs_Tstar"])
for nm, t in [("default", .5), ("Tstar", T_STAR)]:
    cm = confusion_matrix(yte, (p_te >= t).astype(int)); plt.figure(figsize=(4.4, 3.9)); plt.imshow(cm, cmap="Blues")
    for (i, j), v in np.ndenumerate(cm): plt.text(j, i, f"{v:,}", ha="center", va="center", color="white" if v > cm.max()/2 else "black")
    plt.xticks([0, 1], ["Low", "High"]); plt.yticks([0, 1], ["Low", "High"]); plt.xlabel("Predicted"); plt.ylabel("Actual"); plt.title(f"Confusion matrix (T = {t:.4f})"); save(f"fig_confusion_{nm}")
f_, t_, _ = roc_curve(yte, p_te); plt.figure(figsize=(5.2, 4.2)); plt.plot(f_, t_, color=BLUE, lw=2, label=f"AUC = {m05['auc']:.4f}"); plt.plot([0, 1], [0, 1], "--", color=GREY)
plt.xlabel("False positive rate"); plt.ylabel("True positive rate"); plt.title("ROC curve (test set)"); plt.legend(); save("fig_roc")

# %% [markdown]
# ## 7. Ablation and same-split baselines (reviewer 3, comments 10-11)

# %%
abl = {"A. default XGBoost @0.50": None, "B. Bayes-opt @0.50": m05, "C. Bayes-opt @T* (out-of-fold)": mT}
dflt = pipe(xgb({})).fit(Xdev, ydev); abl["A. default XGBoost @0.50"] = metrics(yte, dflt.predict_proba(Xte)[:, 1], .5)
p_pl, p_is = platt.predict_proba(lg(p_te))[:, 1], iso.predict(p_te)
tP = roc_curve(ydev, platt.predict_proba(lg(oof))[:, 1]); tP = float(tP[2][np.argmax(tP[1]-tP[0])])
tI = roc_curve(ydev, iso.predict(oof)); tI = float(tI[2][np.argmax(tI[1]-tI[0])])
abl["D. C + Platt scaling"] = metrics(yte, p_pl, tP); abl["E. C + isotonic"] = metrics(yte, p_is, tI)
base = {"Logistic regression": Pipeline([("s", StandardScaler()), ("c", LogisticRegression(max_iter=2000))]),
        "Random forest": RandomForestClassifier(300, n_jobs=-1, random_state=SEED),
        "SVM (RBF)": Pipeline([("s", StandardScaler()), ("c", SVC(probability=True, random_state=SEED))])}
try:
    from lightgbm import LGBMClassifier; base["LightGBM"] = LGBMClassifier(random_state=SEED, verbose=-1)
except Exception: pass
try:
    from catboost import CatBoostClassifier; base["CatBoost"] = CatBoostClassifier(random_seed=SEED, verbose=0)
except Exception: pass
for n, m in base.items():
    sub = np.arange(len(Xdev)) if n != "SVM (RBF)" else np.random.default_rng(SEED).choice(len(Xdev), min(20000, len(Xdev)), replace=False)
    m.fit(Xdev.iloc[sub], ydev[sub]); pp = m.predict_proba(Xte)[:, 1]
    abl[f"{n} @0.50"] = metrics(yte, pp, .5)
abl_df = pd.DataFrame(abl).T[["accuracy", "precision", "recall", "specificity", "f1", "auc", "brier", "ece"]].round(5)
abl_df.to_csv(f"{OUT}/Table_ablation_baselines.csv"); print(abl_df); R["ablation"] = abl

# %% [markdown]
# ## 8. Repeated stratified CV on the full cleaned data (robustness)

# %%
rcv = RepeatedStratifiedKFold(n_splits=K_FOLDS, n_repeats=3, random_state=SEED)
for sc in ["accuracy", "roc_auc"]:
    s = cross_val_score(pipe(xgb(best)), X, y, cv=rcv, scoring=sc); h = 1.96*s.std()/np.sqrt(len(s))
    R[f"repeated_cv_{sc}"] = dict(mean=float(s.mean()), sd=float(s.std()), ci95=[float(s.mean()-h), float(s.mean()+h)]); print(sc, R[f"repeated_cv_{sc}"])

# %% [markdown]
# ## 9. SHAP (explains the model; it does not prove clinical causality)

# %%
names = ["Age"] + [c for c in feats if c != "Age"]                       # ColumnTransformer puts Age first
Xt = pd.DataFrame(final[:-1].transform(Xte), columns=names)
sv = shap.TreeExplainer(final[-1]).shap_values(Xt)
plt.figure(); shap.summary_plot(sv, Xt, show=False); save("fig_shap_summary")
plt.figure(); shap.summary_plot(sv, Xt, plot_type="bar", show=False); save("fig_shap_bar")
imp = pd.Series(np.abs(sv).mean(0), index=names).sort_values(ascending=False); imp.to_csv(f"{OUT}/Table_shap_importance.csv", header=["mean_abs_shap"]); print(imp)
rng = np.random.default_rng(SEED); rho = []
for _ in range(20):
    i = rng.integers(0, len(sv), len(sv)); rho.append(spearmanr(imp.reindex(names), pd.Series(np.abs(sv[i]).mean(0), index=names)).correlation)
R["shap_bootstrap_spearman_mean"] = float(np.mean(rho)); print("SHAP rank stability (bootstrap Spearman):", R["shap_bootstrap_spearman_mean"])
for f in ["Age", "High_BP", "Chest_Pain"]:
    shap.dependence_plot(f, sv, Xt, show=False); save(f"fig_shap_dependence_{f}")

# %% [markdown]
# ## 10. Calibration ASSESSMENT (Brier, ECE, reliability) and genuine calibration (Platt / isotonic)

# %%
cal = pd.DataFrame({"Brier": [brier_score_loss(yte, p) for p in (p_te, p_pl, p_is)], "ECE": [ece_score(yte, p) for p in (p_te, p_pl, p_is)]},
                   index=["Uncalibrated XGBoost", "Platt scaling", "Isotonic regression"]).round(6)
print(cal); cal.to_csv(f"{OUT}/Table_calibration.csv"); R["calibration"] = cal.to_dict()
plt.figure(figsize=(5.4, 4.8)); plt.plot([0, 1], [0, 1], "--", color=GREY, label="Perfect")
for nm, pp, col in [("Uncalibrated", p_te, BLUE), ("Platt", p_pl, "#2e7d32"), ("Isotonic", p_is, RED)]:
    fy, fx = calibration_curve(yte, pp, n_bins=10, strategy="quantile"); plt.plot(fx, fy, "o-", color=col, label=nm)
plt.xlabel("Mean predicted probability"); plt.ylabel("Observed fraction positive"); plt.title("Reliability diagram (test set)"); plt.legend(); save("fig_calibration")

# %% [markdown]
# ## 11. Five probability bands (fixed limits) - NOT clinically validated

# %%
LIMITS = [0, .2, .4, .6, .8, 1.0000001]; BANDS = ["Band 1 (lowest)", "Band 2", "Band 3", "Band 4", "Band 5 (highest)"]
ADVICE = ["Maintain healthy lifestyle and routine screening.", "Consider lifestyle modification and physician consultation.",
          "Consider clinical evaluation (ECG, blood pressure).", "Consider early specialist consultation.",
          "Prioritise for urgent clinical assessment."]               # ILLUSTRATIVE ONLY - needs cardiologist validation
res = pd.DataFrame({"Actual": yte, "Predicted_at_Tstar": (p_te >= T_STAR).astype(int), "Probability": p_te})
res["Band"] = pd.cut(res.Probability, LIMITS, labels=BANDS, right=False)
res["Illustrative_message"] = res.Band.map(dict(zip(BANDS, ADVICE)))
res.to_csv(f"{OUT}/Test_predictions_with_bands.csv", index=False)
tab = res.groupby("Band", observed=False).agg(Count=("Actual", "size"), Observed_positive_rate=("Actual", "mean")).reindex(BANDS)
tab["Percent"] = (tab.Count / tab.Count.sum() * 100).round(2); print(tab); tab.to_csv(f"{OUT}/Table_band_distribution.csv")
q = pd.qcut(res.Probability, 5, labels=BANDS, duplicates="drop").value_counts().reindex(BANDS)
print("\nFor comparison, QUANTILE bins would give (this is what the old Table 8 looked like - not the fixed bins):\n", q)
fig, ax = plt.subplots(1, 2, figsize=(9.5, 3.8))
ax[0].bar(["B1", "B2", "B3", "B4", "B5"], tab.Count, color=BLUE); ax[0].set(ylabel="Patients", title="(a) Counts")
ax[1].bar(["B1", "B2", "B3", "B4", "B5"], tab.Percent, color=BLUE); ax[1].set(ylabel="Patients (%)", title="(b) Percentages")
for a, v, f in [(ax[0], tab.Count, "{:,}"), (ax[1], tab.Percent, "{:.1f}%")]:
    for i, x in enumerate(v): a.text(i, x, f.format(x), ha="center", va="bottom", fontsize=9)
save("fig_risk_bands")

# %% [markdown]
# ## 12. Literature comparison (descriptive only - other studies use other data and protocols)

# %%
st = ["CNN-LSTM", "HFDT", "CatBoost", "Stacking", "XAI", "CardioRiskNet", "This work*"]
ac = [99.29, 98.30, 98.11, 97.80, 96.07, 98.70, round(100*mT["accuracy"], 2)]
plt.figure(figsize=(8, 4)); b = plt.bar(st, ac, color=[GREY]*6 + [BLUE])
for r_, v in zip(b, ac): plt.text(r_.get_x()+r_.get_width()/2, v+.04, f"{v:.2f}", ha="center", fontsize=9)
plt.ylim(95, 100); plt.ylabel("Reported accuracy (%)"); plt.title("Reported accuracies (different datasets/protocols; not a like-for-like ranking)", fontsize=10); plt.xticks(rotation=20); save("fig_comparison")
json.dump(R, open(f"{OUT}/results.json", "w"), indent=2, default=float)
print("\nAll outputs saved in", OUT)
```
